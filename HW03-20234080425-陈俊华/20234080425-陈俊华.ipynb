```python
# 2.1 理论题数值验算
Cin, Hin, Win = 3, 32, 32
Cout = 16
kh, kw = 5, 5
P, S = 2, 2

# 计算输出高宽
Hout = (Hin + 2 * P - kh) // S + 1
Wout = (Win + 2 * P - kw) // S + 1
print(f"输出特征图尺寸：{Cout} × {Hout} × {Wout}")

# 单个像素点乘次数
dot_times = Cin * kh * kw
print(f"单个输出通道一个像素点乘次数：{dot_times}")
```

```python
import numpy as np

def max_pool2d_np(input_x: np.ndarray, kernel_size: int, stride: int, padding: int):
    N, C, H_in, W_in = input_x.shape
    k = kernel_size
    # 零填充
    x_pad = np.pad(
        input_x,
        pad_width=((0,0), (0,0), (padding,padding), (padding,padding)),
        mode="constant",
        constant_values=0
    )
    H_pad = H_in + 2 * padding
    W_pad = W_in + 2 * padding
    # 输出尺寸
    H_out = (H_pad - k) // stride + 1
    W_out = (W_pad - k) // stride + 1
    out = np.zeros((N, C, H_out, W_out), dtype=np.float32)
    # 滑动窗口
    for n in range(N):
        for c in range(C):
            for h in range(H_out):
                h_start = h * stride
                h_end = h_start + k
                for w in range(W_out):
                    w_start = w * stride
                    w_end = w_start + k
                    window = x_pad[n, c, h_start:h_end, w_start:w_end]
                    out[n, c, h, w] = np.max(window)
    return out

# 测试
if __name__ == "__main__":
    x = np.random.randn(2, 3, 8, 8)
    out = max_pool2d_np(x, kernel_size=3, stride=2, padding=1)
    print("NumPy手动池化输出shape:", out.shape)
```
