```python
# %% [markdown]
# # 2 序列模型
# ## 2.1 理论计算题：一阶马尔可夫+拉普拉斯平滑
# 字符序列："ababc"，词汇表 V={a,b,c}
# 原始转移计数统计：
# 序列相邻对：a->b, b->a, a->b, b->c
# 原始转移计数表 count(x_t, x_{t-1})
# count(b|a)=2, count(a|b)=1, count(c|b)=1，其余转移为0
# 词汇大小 |V|=3，拉普拉斯平滑 p(y|x) = (count(x,y)+1) / (sum_y count(x,y)+|V|)
#
# 1. 计算 p(a|b)
# sum_y count(b,y) = count(a|b)+count(c|b)+count(b|b) = 1+1+0 = 2
# p(a|b) = (1+1)/(2+3) = 2/5 = 0.4
#
# 2. 计算 p(c|b)
# p(c|b) = (1+1)/(2+3) = 2/5 = 0.4

# %% [markdown]
# ## 2.2 编程题：文本预处理函数 preprocess_text(text, n)
# %%
import string
from collections import Counter

def preprocess_text(text, n):
    # 1. 转小写，去除标点，仅保留字母空格
    text_lower = text.lower()
    punc = set(string.punctuation)
    clean_chars = [c for c in text_lower if c not in punc]
    clean_text = "".join(clean_chars)
    # 2. 按空格分词
    tokens = clean_text.split()
    # 3. 构建词汇表：按频率降序，id从0开始
    cnt = Counter(tokens)
    sorted_words = sorted(cnt.keys(), key=lambda w: (-cnt[w], w))
    vocab = {word: idx for idx, word in enumerate(sorted_words)}
    # 4. 滑动窗口生成特征序列(长度n) + 标签
    features = []
    labels = []
    seq_len = len(tokens)
    for i in range(seq_len):
        end = i + n
        if end >= seq_len:
            continue
        feat_window = tokens[i:end]
        label = tokens[end]
        features.append(feat_window)
        labels.append(label)
    return vocab, (features, labels)

# 测试示例
if __name__ == "__main__":
    test_text = "The time machine"
    n = 2
    vocab, (feats, labs) = preprocess_text(test_text, n)
    print("词汇表：", vocab)
    print("特征序列：", feats)
    print("标签序列：", labs)
```
